#Basic Decision-Making statements

```python
#1.Check whether a number is positive or negative.

print(" check if your number is positive or negative ")

n=float(input("Enter a number :"))

if(n>=0):
  print("The number",n,"is Positive")
else:
  print("The number",n,"is Negative")
```

```python
#2.Check whether a number is even or odd.

print(" Check if the number is odd or even ")

n=float(input("Enter a number :"))

if(n%2==0):
  print(n,"is even number")
else:
  print(n,"is odd number")
```

```python
#4.Find the greater of two numbers.

print(" Check which of the two entered values is greater ")

num1=float(input("Enter 1st Number :"))
num2=float(input("Enter 2nd Number :"))

if(num1>num2):
  print(num1,"is greater ")
else:
  print(num2,"is greater")
```

```python
#5.Check whether a student has passed or failed. Passing marks are 40.

print(" Check if the student is passed ")

name=input("Enter student name :")
mark=int(input("Enter the student mark :"))

if(mark>=40):
  print(name,"Result = Pass")
else:
  print(name,"Result = Fail")
```

```python
#6.Check whether a number is divisible by 5.

print(" Check if a number is divisible by 5 ")
num=int(input("Enter a Number : "))

if(num%5==0):
  print("divisible value",num/5)
  print(num,"is divisible by 5")
else:
  print(num,"is not divisible by 5")
```

```python
#8.Display a grade based on the student's marks:
#80–100 → A
#60–79 → B
#40–59 → C
#Below 40 → Fail

mark=int(input("Enter your marks :"))

if(mark>=80 and mark<=100):
  print("grade A")
elif(mark>=60 and mark<=79):
  print("grade B")
elif(mark>=40 and mark<=59):
  print("grade C")
elif(mark<40):
  print("Fail")
```

```python
#9.Check whether a given year is a leap year.

print(" Check whether a entered year is a leap year ")

year=int(input("Enter a year :"))

if(year%4==0):
  print(year,"is a leap year")
elif(year%400==0):
  print(year,"is a leap year")
else:
  print(year,"is not a leap year")
```
