#Loop-Based questions

Use for and/or while loops.

```python
#1.Print numbers from 1 to 10.

for i in range(1,10+1,1):
  print(i)
```

```python
#2.Print numbers from 10 to 1.

i=10
while(i>=1):
  print(i)
  i=i-1
```

```python
#3.Print all even numbers from 1 to 20.

for i in range(0,20+1,2):
  print(i) #even numbers will be printed
```

```python
#4.Print all odd numbers from 1 to 20.

i=1
while(i<=20):
  print(i) #odd numbers will be printed
  i=i+2
```

```python
#5.Take two numbers from the user and print all even numbers between them.

num1=int(input("Enter 1st number: "))
num2=int(input("Enter 2nd number: "))

for i in range(num1,num2+1,1):
  if(i%2==0):
    print(i)
```

```python
#6.Take two numbers from the user and print all odd numbers between them.

num1=int(input("Enter 1st number: "))
num2=int(input("Enter 2nd number: "))

i=num1
while(i<=num2):
  if(i%2==1):
    print(i)
  i=i+1
```

```python
#7.Take a number from the user and print its multiplication table.

num=int(input("Enter the multiplication table number: "))

value=0
for i in range(num*1,num*10+1,num):
  value=value+1
  print(value, "x", num, "=", i)
```

```python
#8.Find the sum of numbers from 1 to 10.

sum=0
for i in range(1,10+1,1):
  sum=sum+i
print("Sum of 1 - 10 is =",sum)
```

```python
#9.Take two numbers and find thesum of all numbers between them

num1=int(input("Enter 1st number: "))
num2=int(input("Enter 2nd number: "))

sum=0
for i in range(num1,num2+1,1):
  sum=sum+i
print("Sum of range is =",sum)
```

```python
#10.Take two numbers and print all numbers divisible by 6 between them.

print( "Enter range to get numbers divisible by 6 in that range ")

num1=int(input("Enter 1st number : "))
num2=int(input('Enter 2nd number : '))

for i in range(num1,num2+1,1):
  if(i%6==0):
    print(i)
```

```python
#11.Take two numbers and print all numbers that are even and divisible by 6.

num1=int(input("Enter 1st number: "))
num2=int(input("Enter 2nd number: "))

for i in range(num1,num2+1,1):
  if(i%6==0 and i%2==0):
    print(i)
```

```python
#12.Take two numbers and print all numbers that are odd and divisible by 3.

num1=int(input("Enter 1st value: "))
num2=int(input("Enter 2nd value: "))

for i in range(num1,num2+1,1):
  if(i%3==0 and i%2==1):
    print(i)
```

```python
#13.Find the factorial of a given number.

print(" Finding factorial of a number ")

num=int(input("Enter a number : "))

fact=1
for i in range(1,num+1,1):
  fact=fact*i
print("Factorial of",num,"is =",fact)
```
