1.)Create a menu-driven program to perform the following operations:

1 → Addition
2 → Subtraction
3 → Multiplication
4 → Division

```python
print(" CALCULATOR ")

print("1 → Addition")
print("2 → Subtraction")
print("3 → Multiplication")
print("4 → Division")

num1=int(input("Enter the 1st number :"))
num2=int(input("Enter the 2nd number :"))

ch=int(input("Enter your choice :"))

if(ch==1):
  num3=num1+num2
  print("added value is ",num3)
elif(ch==2):
  num3=num1-num2
  print("subracted value is ",num3)
elif(ch==3):
  num3=num1*num2
  print("multiplied value is ",num3)
elif(ch==4):
  num3=num1/num2
  print("divided value is ",num3)
else:
  print("Invalid choice!")
```

2.)Create a menu-driven program for a student:
1 → Display Student Name
2 → Display Student Marks
3 → Display Student Grade

```python
print(" STUDENT DETAILS ")
print("1 → Display Student Name")
print("2 → Display Student Marks")
print("3 → Display Student Grade")

name=input("Enter student name: ")
mark=int(input("Enter student Mark: "))
ch=int(input("Enter the choice: "))

if(ch==1):
  print("Student name:",name)
elif(ch==2):
  print("Student mark:",mark)
elif(ch==3):
  if(mark>=80 and mark<=100):
    print("grade A")
  elif(mark>=60 and mark<=79):
    print("grade B")
  elif(mark>=40 and mark<=59):
    print("grade C")
  elif(mark<40):
    print("Fail")
```

3.)Create a menu-driven program for a basic banking system:
1 → Check Balance
2 → Deposit
3 → Withdraw

```python
print(" Basic Banking SYstem ")

print("1 → Check Balance")
print("2 → Deposit")
print("3 → Withdraw")
print("4 → Exit")

current_balance=50000
ch=0
while(ch<4):
  ch=int(input("Enter your choice: "))
  if (ch==1):
    print("Your current balance is",current_balance)
  elif(ch==2):
    deposit=int(input("Enter the deposit amount in Rs:"))
    current_balance=current_balance+deposit
    print("Amount deposited succesfully","Current balance:",current_balance)
  elif(ch==3):
      withdraw=int(input("Enter withdrawal amount:"))
      current_balance-=withdraw
      print("Amount withdrawal successful!","Current balance",current_balance)
  elif(ch==4):
    print("Thank you!")
  else:
    print("Invalid choice")
    break

  ch=ch+1
```

4.)Create a menu-driven program to calculate:
1 → Area of Rectangle
2 → Area of Circle
3 → Area of Square

```python
print("1 → Area of Rectangle")
print("2 → Area of Circle")
print("3 → Area of Square")

ch=int(input("Enter your choice : "))

if(ch==1):
  width=int(input("Enter the width of rectangle :"))
  length=int(input("Enter the length of rectangle : "))
  area=width*length
  print("The area of rectangle is",area)
elif(ch==2):
  radius=int(input("Enter the cirle radius : "))
  area=3.14*radius**2
  print("The area of circle is",area)
elif(ch==3):
  sides=int(input("Enter the side value :"))
  area=sides**2
  print("The area of square is",area)
else:
  print("Invalid choice!")
```

5.)Create a menu-driven program to convert:
1 → Celsius to Fahrenheit
2 → Fahrenheit to Celsius

```python
print(" Converting temperatures program ")

print("1 → Celsius to Fahrenheit")
print("2 → Fahrenheit to Celsius")

ch=int(input("Enter your choice "))

if(ch==1):
  c=int(input("Enter the celcius value:"))
  f=(c*9/5)+32
  print(c,"°C is =",f,"°F")
elif(ch==2):
  f=int(input("Enter the ferenheit value:"))
  c=(f-32)*5/9
  print(f,"°F is =",round(c,21),"°C")
else:
  print("Invalid choice!")
```
