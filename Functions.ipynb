```python
#6.Create a function to find the greater of two numbers.

def greater(num1,num2):

  if(num1>num2):
    print(num1,"is greater than",num2)
  elif(num2>num1):
    print(num2,"is greater than",num1)
  elif(num1==num2):
    print("Both are equal")


num1=int(input("Enter 1st number "))
num2=int(input("Enter 2nd number "))

greater(num1,num2)
```

```python
#8.Create a function to calculate the factorial of a number.

def fact(num):
  fact=1
  for i in range(1,num+1,1):
    fact=fact*i
  print("Factorial of",num,"is",fact)

num=int(input("Enter a number :"))
fact(num)
```

```python
#10.Create a function that accepts marks and returns the student's grade.

def grade(mark):

  if (mark>=80 and mark<=100):
    grade="Grade = A"
  elif(mark>=60 and mark<=79):
    grade="Grade = B"
  elif(mark>=40 and mark<=59):
    grade="Grade = C"
  else:
    grade="Grade = Fail"
  return grade

mark=int(input("Enter your marks :"))
grade(mark)
```

```python
#2.Create a function that accepts a list and returns the sum of all elements

def sum_of_list(my_list):
  total=0
  for i in range(0,len(my_list),1):
    total=total+my_list[i]
  return total

mark=[96,83,89,76,57]

print("Total marks=",sum_of_list(mark))
```

```python
#4.Create a function that accepts a list of marks and returns the number of passed students.

def results(my_list):
  counting=0
  for i in range(0,len(my_list),1):
    if (my_list[i]>=40):
     counting=counting+1
  return counting

mark=[96,83,32,89,24,76,57,39,40]

print("PASSED STUDENTS = ",results(mark))
```

```python
#5.Create a function that accepts a list of marks and returns the number of failed students.

def result(my_list):
  counting=0
  for i in range(0,len(my_list),1):
    if(my_list[i]<40):
      counting+=1
  return counting

mark=[96,83,32,89,24,76,57,39,40]

print("FAILED STUDENTS =",result(mark))
```

```python
#6.Create a function that accepts a list and returns a new list containing only even numbers.

def even(my_list):
  new_list=[]
  for i in range(0,len(my_list),1):
   if(my_list[i]%2==0):
    new_list.append(my_list[i])
  return new_list

numbers=[45,27,99,41,65,83,96,32,89,24,76,57,39,40,64]

print("Even number list",even(numbers))
```

C. Functions with Dictionaries

```python
#1.Create a function that accepts a student dictionary and displays the student's name and marks.

def display(my_dictionary):
  for keys,values in my_dictionary.items():
    print(keys,":",values)

Students={"Menaka":96,"Jeigiri":80,"Mithuna":75,"Raje":86,"Velu":86,"Karan":79,"Nikhil":76}

display(Students)
```

```python
#2.Create a function that accepts a dictionary of student marks and returns the highest marks.

def display_max(my_dictionary):
  highest_mark=max(my_dictionary.values())
  for name,mark in my_dictionary.items():
    if (mark==highest_mark):
      print(name,"has the Highest mark = ",mark)

Students={"Menaka":96,"Jeigiri":80,"Mithuna":75,"Raje":86,"Velu":86,"Karan":79,"Nikhil":76}

display_max(Students)
```

```python
#3.Create a function that accepts a dictionary of students and marks and returns the number of passed students.

def display_pass(my_dictionary):
  counting=0
  for i in my_dictionary.values():
    if (i>=40):
     counting+=1
  return counting

Students={"Menaka":96,"Jeigiri":80,"Mithuna":75,"Raje":86,"Velu":86,"Karan":79,"Nikhil":76}

print("Passed students = ",display_pass(Students))
```

```python
#4.Create a function that accepts a product-price dictionary and returns the total price of all products.

def total_price(my_dictionary):
  total=0
  for i in my_dictionary.values():
    total=total+i
  return total

product={"Milk":25,"Biscuit":20,"Bread":40,"Rice bag":300,"Oil":70}

print("Total price=",total_price(product))
```
