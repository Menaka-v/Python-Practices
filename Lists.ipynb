#Lists

```python
#Using the following list for Questions 1–10:

marks = [45, 67, 32, 89, 56, 28, 76, 91, 35, 62]
```

```python
#1.Find the length of the list using len().

print("the length of list is =",len(marks))
```

```python
#2.Find the total marks using sum().

print("Sum of list is =",sum(marks))
```

```python
#3.Find the highest marks using max().

print("Max mark =",max(marks))
```

```python
#4.Find the lowest marks using min().

print("Minimum mark =",min(marks))
```

```python
#5.Sort the marks in ascending order.

marks.sort()
print(marks)
```

```python
#6.Sort the marks in descending order.

marks.sort(reverse=True)
print(marks)
```

```python
#7.Count how many students scored more than 50.

count=0
for i in range(0,len(marks),1):
  if (marks[i]>50):
    count=count+1
print(count)
```

```python
#8.Count how many students passed, assuming passing marks are 40.

count=0
for i in range(0,len(marks),1):
  if (marks[i]>=40):
    count=count+1
print(count,"students are passed")
```

```python
#9.Count how many students failed, assuming passing marks are 40.

count=0
for i in range(0,len(marks),1):
  if(marks[i]<40):
    count+=1
print(count,"students are failed")
```

```python
#10.Calculate the average marks of all students.


avg=sum(marks)/len(marks)

print(avg,"is the average mark of all students")
```

```python
#5.Given a list of numbers, count how many numbers are even and odd.

numbers=[13,12,48,879,621,485,93,76,65,795,24,16,79,330]

count1=0
count2=0
for i in range(0,len(numbers),1):
  if(numbers[i]%2==0):
    count1=count1+1
  elif(numbers[i]%2==1):
    count2=count2+1

print(count1,"Even numbers")
print(count2,"Odd numbers")
```

```python
#8.Given a list of student marks, create two lists: pass_students,fail_students

marks=[12,13,50,89,99,76,89,49,73,61,48,76,17,69,33,80,76,81]

pass_students=[]
fail_students=[]
for i in range(0,len(marks),1):
  if(marks[i]>=40):
    pass_students.append(marks[i])
  elif(marks[i]<40):
    fail_students.append(marks[i])


print("passed students mark list",pass_students)
print("failed students mark list",fail_students)
```
