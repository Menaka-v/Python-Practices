```python
#Use the following dictionary for Questions 1–8:
student = {
    "name": "Menaka",
    "age": 20,
    "marks": 75,
    "city": "Tiruppur"}
```

```python
#1.Print the student's name.

print("Student name is ",student["name"])
```

```python
#2.Print the student's marks.

print("mark =",student["marks"])
```

```python
#3.Update the student's marks to 85.

student["marks"]=85

print("updated mark =",student["marks"])
```

```python
#4.Add a new key called "course"

student["course"]="B.com CA"

print(student)
```

```python
#5.Add an "email" key.

student["email"]="menaka26@gmail.com"

print(student)
```

```python
#6.Remove the "age" key.

del student["age"]

print(student)
```

```python
#7.Print all keys using keys().

print(student.keys())
```

```python
#8.Print all values using values().

for i in student.values():
  print(i)

print(student.values())
```

Additional dictionary practice

```python
#9.Create a dictionary containing 5 students and their marks. Find the student with the highest marks.

student={"Menaka":96,
         "Jeigiri":92,
         "Mithuna":85,
         "Raje":80,
         "Velu":76}

for key,value in student.items():
  if value==max(student.values()):
    print(key,":",value)
```

```python
#10.Create a dictionary containing 5 products and their prices. Find the most expensive product.

product={"milk":25,
         "biscuit":20,
         "bread":40,
         "rice bag":300,
         "oil":70}

for key,value in product.items():
  if value==max(product.values()):
    print("Most expensive product =",key)

```

```python
#11.Create a dictionary containing student names and marks. Count how many students passed.


student={"Menaka":96,
         "Jeigiri":92,
         "Mithuna":85,
         "Raje":80,
         "Velu":76,
         "pixel":32}

count1=0

for value in student.values():
  if (value>=40):
    count1=count1+1

print("passed =",count1)
```

```python
#14.Create a dictionary of products and prices. Calculate the total value of all products.

product={"milk":25,
         "biscuit":20,
         "bread":40,
         "rice bag":300,
         "oil":70}
total=0
for i in product.values():
  total=total+i
print("Total value =",total)
```
