#Tuples

```python
#1.Create a tuple containing 5 numbers and print it.

num=(1,2,3,4,5)

print(num)
```

```python
#2.Print the first element of a tuple.

fruits=("apple","pineapple","banana","cherry","orange")

print("first element :",fruits[0])
```

```python
#3.Print the last element of a tuple.

fruits=("apple","pineapple","banana","cherry","orange")

print("last element :",fruits[-1])
```

```python
#4.Find the length of a tuple using len().

fruits=("apple","pineapple","banana","cherry","orange")

print("length of tuple :",len(fruits))
```

```python
#5.Check whether a particular value exists in a tuple.

check=input("enter a fruit name :")

check=check.lower()

if check in fruits:
  print(check,"exists in fruits!")
else:
  print(check,"doesn't exists in fruits!")
```

```python
#6.Count how many times a value occurs using count().

fruits=("apple","pineapple","banana","cherry","orange","pineapple")

print(fruits.count("pineapple"))
```

```python
#7.Find the index of a value using index().

fruits=("apple","pineapple","banana","cherry","orange")

print(fruits.index("cherry"))
```

```python
#8.Find the maximum value using max().

fruits_price=(50,60,45,70,65)

print("max fruit price :",max(fruits_price))
```

```python
#9.Find the minimum value using min().

fruits_price=(50,60,45,70,65)

print("min fruit price:",min(fruits_price))
```

```python
#10.Find the total of tuple values using sum().

fruits_price=(50,60,45,70,65)

print("total price:",sum(fruits_price))
```
